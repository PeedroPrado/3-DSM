Segue os exemplos, com biblioteca

```
import numpy as np

u = np.array([4, 3, -1])
v = np.array([1, 0, 2])

print("u =", u)
print ("v = ", v)

soma = u + v
print(f"Soma: {soma}")

alpha = 4
produto_escalar = alpha * v #multiplicação por vetor por escalar
print(f"Multiplicação escalar: {produto_escalar}")

## Norma Euclidiana mede o "comprimento" do vetor
norma_v = np.linalg.norm(v)
print(f"Norma de v: {norma_v: .4f}")

## Distância Euclidiana entre dois vetores é a norma da diferença entre eles
distancia = np.linalg.norm(u - v)
print(f"Distância entre u e v: {distancia: .4f}")

## Produto escalar usando np.dot
produto_interno = np.dot(u, v)
print(f"Produto interno (dot): {produto_interno}")

#Produto escalar usando o operador @
produto_interno2 = u @ v
print(f"Produto interno (@)? {produto_interno2}")

# Produto escalar usando a multiplação componente a componente e soma

produto_manual = np.sum(u * v)
print(f"Produto interno (manual): {produto_manual}")
```

Norma Euclidiana: A norma Euclidiana mede o “comprimento” do vetor.

```
import numpy as np

v = np.array([1, 0, 2])
norma_v = np.linalg.norm(v) #O linalg é um submódulo da biblioteca numpy em Python, que significa "álgebra linear". Ele fornece um conjunto de funções para realizar operações de álgebra linear
print(f"Norma de v: {norma_v: 4f}")
```

Distância entre Dois Vetores


> A distância Euclidiana entre dois vetores é a norma da diferença entre eles.







Observação: O linalg é um submódulo da biblioteca numpy em Python, que significa "álgebra linear". Ele fornece
um conjunto de funções para realizar operações de álgebra linear, como:
• Cálculo de normas: Como a função **np.linalg.norm()** que usamos para calcular a norma de um vetor;
• Cálculo de autovalores e autovetores: **np.linalg.eig**();
• Inversão de matrizes: **np.linalg.inv();**
• Determinantes: **np.linalg.det(). **

```
import numpy as np

u = np.array([4, 3, -1])
v = np.array([1, 0, 2])

distancia = np.linalg.norm(u - v)
print(f"Distância entre u e v: {distancia: .4f}")
```

Produto Interno (Produto Escalar)

O produto interno combina duas listas de números, multiplicando componentes e somando os resultados.

```
import numpy as py

u = np.array([4, 3, -1])
v = np.array([1, 0, 2 ])

#Usando o dot
produto_interno3 = np.dot(u,v)
print(f"Produto interno (dot): {produto_interno3}")

#Usando o operdador @
produto_interno4 = u @ v
print(f"Produto interno (@): {produto_interno4}")

#multiplicação componente a componente e soma

produto_manual2 = np.sum (u * v)
print(f"Produto interno (manual): {produto_manual2: .4f}")
```
