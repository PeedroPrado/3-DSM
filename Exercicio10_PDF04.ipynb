# Exercício 10 – Verificação computacional com NumPy (opcional)

b) Construir 𝑃 e 𝐷 a partir da saída do NumPy.
c)
Verificar numericamente que 𝐴 ≈ 𝑃𝐷𝑃−1.

A =[ 4 1
     1 4 ]

```python
import numpy as np
A = np.array([[4, 1],
[1, 4]])
# 1. Autovalores e autovetores
vals, vecs = np.linalg.eig(A)
print("Autovalores:", vals)
print("Autovetores (colunas de P):")
print(vecs)
# 2. Montando P e D
P = vecs
D = np.diag(vals)
# 3. Reconstrução de A
A_rec = P @ D @ np.linalg.inv(P)
print("A reconstruída:")
print(A_rec)
# Verificação
print("Diferença A - A_rec:")
print(A - A_rec)
```
