# Determinante de Matrizes 2×2

```python
import numpy as np

A = np.array([[3, 2],
              [1, 4]])

det_A_manual = A[0,0]*A[1,1] - A[0,1]*A[1,0] # Cálculo manual
det_A_numpy = np.linalg.det(A) # usando NumPy

print("Determinante (manual):", det_A_manual)
print("Determinante (NumPy): ", det_A_numpy)
```

# Determinante de Matrizes 3×3

```python
A = np.array([[1, 2, 3],
              [0, 1, 4],
              [5, 6, 0]])
# Cálculo manual (Sarrus)
det_manual = (
    A[0,0]*A[1,1]*A[2,2] +
    A[0,1]*A[1,2]*A[2,0] +
    A[0,2]*A[1,0]*A[2,1] -
    A[0,2]*A[1,1]*A[2,0] -
    A[0,0]*A[1,2]*A[2,1] -
    A[0,1]*A[1,0]*A[2,2]
)
det_numpy = np.linalg.det(A) # usando NumPy
print("Determinante (manual via Sarrus):", det_manual)
print("Determinante (NumPy):            ", det_numpy)
```

# Propriedades Fundamentais do Determinante

```python
A = np.array([[2, 1],
[3, 4]])
B = A[[1,0]]  # Troca de linhas
print(f"B:{B}")
print("det(A) =", np.linalg.det(A))
print("det(B) =", np.linalg.det(B))
```

Multiplicar uma linha por 𝑘 multiplica o determinante por �

```python
A = np.array([[2, 1],
              [3, 4]])
B = A.copy()
B[1] = 5 * B[1]   # Multiplica a segunda linha por 5
print(f"B:{B}")
print("det(A)      =", np.linalg.det(A))
print("det(5·L2(A)) =", np.linalg.det(B))
print("5 × det(A)  =", 5 * np.linalg.det(A))
```

Somar um múltiplo de uma linha a outra não altera o determinante.

```python
import numpy as np
A = np.array([[2, 1],
              [3, 4]])
B = A.copy()
B[1] = B[1] + 10 * B[0]  # L2 ← L2 + 10·L1
print(f"B:{B}")
print("det(A) =", np.linalg.det(A))
print("det(B) =", np.linalg.det(B))
```

Determinante de matrizes triangulares:

```python
import numpy as np
A = np.array([[2, 3, 1],
              [0, 5, 4],
              [0, 0, 7]])
det_numpy = np.linalg.det(A)
det_manual = A[0,0] * A[1,1] * A[2,2] # 2 * 5 * 7
print("Determinante (manual):", det_manual)
print("Determinante (NumPy): ", det_numpy)
```

Compatibilidade com o produto matricial:

```python
import numpy as np
A = np.array([[2, 1],
              [3, 4]])
B = np.array([[1, 2],
              [0, 5]])
AB = A @ B  # Produto matricial
det_A  = np.linalg.det(A)
det_B  = np.linalg.det(B)
det_AB = np.linalg.det(AB)
print("det(A)  =", det_A)
print("det(B)  =", det_B)
print("det(AB) =", det_AB)
print("det(A)*det(B) =", det_A * det_B)
```

Determinante e inversa:

```python
import numpy as np
A = np.array([[4, 2],
              [3, 1]])

det_A = np.linalg.det(A)

print("det(A) =", det_A)

# Verifica se A é inversível
if abs(det_A) > 1e-12:
    A_inv = np.linalg.inv(A)
    det_A_inv = np.linalg.det(A_inv)

    print("A é inversível.")
    print("det(A^{-1}) =", det_A_inv)
    print("1/det(A)   =", 1/det_A)
else:
    print("A não é inversível (determinante zero).")
```

```python
import matplotlib.pyplot as plt
# Vetores base do plano
v1 = np.array([1, 0])
v2 = np.array([0, 1])
C = np.array([[1, 2],
[2, 4]])
v1_t = C @ v1
v2_t = C @ v2
plt.figure(figsize=(3, 4))
plt.arrow(0,0,*v1, width=0.03, color="red")
plt.arrow(0,0,*v2_t, width=0.03, color="blue")
plt.arrow(0,0,*v1_t, width=0.03, color="magenta")
plt.xlim(-1,3)
plt.ylim(-1,5)
plt.gca().set_aspect('equal')
plt.title("Vetores base sob matriz singular C")
plt.grid(True)
plt.show()
```

Autovalores em Matrizes 2×2

```python
import numpy as np

A = np.array([[4, 2],
              [1, 3]])

vals, vecs = np.linalg.eig(A)

print("Autovalores:")
print(vals)
print("\nAutovetores (colunas):")
print(vecs)

# Verificação do conceito Av = λv
v = vecs[:, 0]
lambda_ = vals[0]

print("\nAv:")
print(A @ v)
print("\nλv:")
print(lambda_ * v)
```

Interpretação Geométrica no Plano (R²)

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_eigen_verification():
    # 1. Definição da Matriz A
    # Usamos uma matriz simétrica para garantir autovetores ortogonais (mais fácil de ver)
    A = np.array([[3, 1],
                  [1, 3]])

    # 2. Cálculo dos Autovalores e Autovetores
    vals, vecs = np.linalg.eig(A)

    # Dica de ML: Ordenar do maior para o menor autovalor (importante para PCA)
    idx = vals.argsort()[::-1]
    vals = vals[idx]
    vecs = vecs[:, idx]

    # Configuração do gráfico
    fig, ax = plt.subplots(figsize=(5, 5))
    # Cores para diferenciar os pares
    colors = ['#FF0000', '#0000FF'] # Vermelho e Azul
    origin = [0, 0]
    print(f"Matriz A:\n{A}\n")

    # 3. Loop para plotar cada autovetor e sua transformação
    for i in range(len(vals)):
        lambda_val = vals[i]
        v = vecs[:, i]      # Autovetor original (normalizado pelo NumPy)
        Av = A @ v          # Vetor transformado (esticado por lambda)

        color = colors[i]
        print(f"--- Par {i+1} ---")
        print(f"Autovalor (lambda): {lambda_val:.2f}")
        print(f"Autovetor (v): {v}")
        print(f"Transformado (Av): {Av}")
        print(f"Verificação: Av == lambda * v? {np.allclose(Av, lambda_val * v)}\n")

        # A) Plotar o Vetor transformado (Av) -> Seta Grossa e Transparente
        # Isso representa o resultado da transformação (esticamento)
        ax.quiver(*origin, Av[0], Av[1], color=color, alpha=0.3,
                  angles='xy', scale_units='xy', scale=1,
                  width=0.025, label=f'Transformado Av (λ={lambda_val:.1f})' if i==0
else "")

        # B) Plotar o Autovetor original (v) -> Seta Fina e Sólida
        # Isso representa a direção original
        ax.quiver(*origin, v[0], v[1], color=color, alpha=1.0,
                  angles='xy', scale_units='xy', scale=1,
                  width=0.008, label=f'Original v' if i==0 else "")

        # Adicionar anotação de texto para identificar o autovalor
        offset = 0.2
        ax.text(Av[0] + offset, Av[1], f"λ={lambda_val:.0f}", color=color,
fontsize=12, fontweight='bold')
        # 4. Adicionar um vetor "Comum" (NÃO autovetor) para contraste
    # Para mostrar que um vetor qualquer rodaria e sairia do alinhamento
    v_random = np.array([1, 0])
    Av_random = A @ v_random

    # Plota vetor não-eigen transformado (tracejado) para mostrar que ele gira
    ax.quiver(*origin, Av_random[0], Av_random[1], color='gray', alpha=0.5,
              angles='xy', scale_units='xy', scale=1, width=0.005, linestyle='--')
    ax.text(Av_random[0], Av_random[1], "Vetor comum\n(mudou direção)", color='gray',
fontsize=9)

    # Ajustes finais do gráfico
    limit = 4.5
    ax.set_xlim(-limit, limit)
    ax.set_ylim(-limit, limit)
    ax.axhline(0, color='black', linewidth=0.5)
    ax.axvline(0, color='black', linewidth=0.5)
    ax.grid(True, linestyle='--', alpha=0.6)
    ax.set_aspect('equal')
    ax.set_title(f"Prova Visual: Autovetores não mudam de direção\n(Apenas esticam por um fator λ)")

    # Criar legenda personalizada para explicar a sobreposição
    from matplotlib.lines import Line2D
    custom_lines = [Line2D([0], [0], color='gray', lw=8, alpha=0.3,
label='Transformado (Av)'),
                    Line2D([0], [0], color='gray', lw=2, label='Original (v)')]
    ax.legend(handles=custom_lines, loc='upper left')

    plt.show()
plot_eigen_verification()
```
