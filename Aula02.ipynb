# Aula 2:  Combinação Linear, Span, Independência Linear e Mudança de Base


> Exemplos em Python:
1. Interpretação Geométrica: Criando deslocamentos



```python
import numpy as np
import matplotlib.pyplot as plt

v1 = np.array([1, 2])
v2 = np.array([3, 1])

#Geração de combinações lineares
a_vals = np.linspace(-3, 3, 10) # do -3 ao 3 com 10 amostras
pontos_span = []

#Calculo dos pontos
for a1 in a_vals:
  for a2 in a_vals:
    ponto = a1 * v1 + a2 * v2
    pontos_span.append(ponto)

pontos_span = np.array(pontos_span)

#Plotagem
plt.figure(figsize= (5, 5))
plt.scatter(pontos_span[:, 0], pontos_span[:, 1],
            s=10, alpha=0.4, label='Combinações lineares (span)')

plt.quiver(0, 0, v1[0], v1[1],
           angles='xy', scale_units='xy', scale=1,
           color='blue', label=f'v1={v1}')

plt.quiver(0, 0, v2[0], v2[1],
           angles='xy', scale_units='xy', scale=1,
           color='green', label=f'v2={v2}')

lim = np.max(np.abs(pontos_span)) + 1
plt.xlim(-lim, lim)
plt.ylim(-lim, lim)

plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)

plt.grid(True, linestyle='--', alpha=0.4)
plt.gca().set_aspect('equal', adjustable='box')
plt.legend()
plt.title('Visualização do Span: Todas as Combinações Lineares de v1 e v2')

plt.show()
```

# Exemplo de span de vetores colineares:

O colapso dimensional: Ao rodar o código,
vemos que, embora tenhamos dois vetores e
dois loops de repetição variando os pesos,
todos os pontos laranja caem perfeitamente
sobre uma única linha diagonal.
2. Redundância: O vetor verde 𝑣2 aponta
exatamente na mesma direção que o azul 𝑣1,
apenas é mais longo. Ele não oferece uma
"nova direção" para explorar o plano 𝑅2.
3. Conceito de span: Isso prova visualmente que

s𝑝𝑎𝑛{𝑣1,𝑣2} é uma reta.

```python
import numpy as np
import matplotlib.pyplot as plt

v1 = np.array([1, 2])
v2 = np.array([2 ,4])

#Combinações lineares (amostragem)
a_vals = np.linspace(-3, 3, 20)
pontos_span = []

for a1 in a_vals:
  for a2 in a_vals:
    ponto = a1 * v1 + a2 * v2
    pontos_span.append(ponto)

pontos_span = np.array(pontos_span)
# 3. Plotagem
plt.figure(figsize=(5, 5))
# Plotando os pontos gerados (Span)
plt.scatter(pontos_span[:, 0], pontos_span[:, 1],
            s=20, alpha=0.5, color='orange', label='Span (Combinações)')
# Plotando os vetores originais
plt.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1,
           color='blue', label=f'v1={v1}', width=0.015)
plt.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1,
           color='green', label=f'v2={v2}', width=0.010)
# Configurações visuais
lim = 10
plt.xlim(-lim, lim)
plt.ylim(-lim, lim)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.grid(True, linestyle='--', alpha=0.4)
plt.gca().set_aspect('equal', adjustable='box')
plt.legend()
plt.title('Span de vetores colineares: O Plano colapsa numa Reta')
plt.show()
```

#  Visualização em Python: O Teorema da Dimensionalidade

```python
import numpy as np
import matplotlib.pyplot as plt

v1 = np.array([2, 1])
v2 = np.array([-1, 1])
#V3 criado como combinação de v1 e v2
v3 = 1 * v1 + 2 * v2

origin = [0], [0]

plt.figure(figsize=(5, 4))

plt.quiver(*origin, v1[0], v1[1], color='black', scale=1, scale_units='xy',
           angles='xy', label = 'v1 (Base)' )
plt.quiver(*origin, v2[0], v2[1], color='red', scale=1, scale_units='xy',
angles='xy', label='v2 (Base)')
plt.quiver(*origin, v3[0], v3[1], color='green', scale=1, scale_units='xy',
angles='xy', label='v3 (Dependente)')
# Mostra que v3 é alcançável andando 1x em v1 e depois 2x em v2
plt.plot([v1[0], v3[0]], [v1[1], v3[1]], 'r--', alpha=0.5, label='Caminho v1 -> v3')
# Ajustes do gráfico
plt.xlim(-3, 3)
plt.ylim(-1, 4)
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend(loc='upper left')
plt.title("Dependência Linear em 2D\nO v3 é redundante, pois v3 = 1*v1 + 2*v2")
plt.show()
```

# Visualização Geométrica

Explicação do Gráfico:
• O vetor 𝑣 é o destino.
• Os vetores 𝑏1 e 𝑏2 são a nossa base.
• As linhas pontilhadas mostram o "caminho"
seguindo as instruções da base.
• O código calcula exatamente quanto
precisamos "esticar" o azul e o verde para
chegar ao destino. Esses valores de
esticamento (𝑐1,𝑐2) são as coordenadas.

```python
import numpy as np
import matplotlib.pyplot as plt

v = np.array([4, 3]) # Vetor alvo (em coordenadas padrão)
# Nova Base (Base B)
b1 = np.array([2, 1])   # Vetor Base 1
b2 = np.array([-1, 1])  # Vetor Base 2
# 2. Calculando as Coordenadas (Resolvendo o Sistema Linear)

# Montamos a Matriz de Transformação com b1 e b2 como COLUNAS
# Sistema: c1*b1 + c2*b2 = v  --> Matriz * c = v
M_base = np.column_stack((b1, b2))
# Resolvemos para encontrar c (as coordenadas na nova base)
# c = [c1, c2]
coords = np.linalg.solve(M_base, v)
c1, c2 = coords
print(f"Vetor original (Padrão): {v}")
print(f"Novas Coordenadas na Base B: c1={c1:.2f}, c2={c2:.2f}")
print(f"Verificação: {c1:.2f}*{b1} + {c2:.2f}*{b2} = {c1*b1 + c2*b2}")
# 3. Visualização
origin = [0], [0]
plt.figure(figsize=(5, 5))
# Plotando a Base B
plt.quiver(*origin, b1[0], b1[1], color='blue', scale=1, scale_units='xy',
angles='xy', label=f'b1={b1}')
plt.quiver(*origin, b2[0], b2[1], color='green', scale=1, scale_units='xy',
angles='xy', label=f'b2={b2}')
# Plotando o vetor V
plt.quiver(*origin, v[0], v[1], color='red', scale=1, scale_units='xy',
angles='xy', label=f'v={v}')
# Visualizando a decomposição (O Caminho)
# Primeiro andamos c1 na direção de b1
step1 = c1 * b1
plt.quiver(*origin, step1[0], step1[1], color='blue', scale=1, scale_units='xy',
angles='xy', alpha=0.3)
# Depois, a partir de onde paramos, andamos c2 na direção de b2
plt.quiver(step1[0], step1[1], c2*b2[0], c2*b2[1], color='green', scale=1,
scale_units='xy', angles='xy', alpha=0.3, width=0.005)
# Linhas pontilhadas para fechar o paralelogramo
plt.plot([0, step1[0]], [0, step1[1]], 'b--', alpha=0.4, label=f'Componente b1 ({c1:.1f} passos)')
plt.plot([step1[0], v[0]], [step1[1], v[1]], 'g--', alpha=0.4,
label=f'Componente b2 ({c2:.1f} passos)')
plt.xlim(-2, 6)
plt.ylim(-1, 6)
plt.grid(True, linestyle=':', alpha=0.6)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend()
plt.title(f"Coordenadas de v na Nova Base:\n v = {c1:.1f} * b1 + {c2:.1f} * b2")
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

# Mudança de Base
  Definição Formal da Mudança de Base

```python
import numpy as np
import matplotlib.pyplot as plt
# 1. Definição do Vetor e da Nova Base (conforme o texto)
v = np.array([4, 3])
# Vetor no sistema padrão
c1 = np.array([2, 1])   # Primeiro vetor da nova base
c2 = np.array([-1, 1])  # Segundo vetor da nova base
# 2. Montagem da Matriz de Mudança de Base
# A matriz P_C deve ter os vetores da base C como COLUNAS
#P_C = [[2, -1], [1, 1]]
P_C = np.column_stack((c1, c2))
# 3. Resolução do Sistema Linear
# Queremos encontrar d (vetor de coordenadas novas) tal que: P_C * d = v
d = np.linalg.solve(P_C, v)
d1, d2 = d
# Exibição dos resultados numéricos
print("=== Resultados da Mudança de Base ===")
print(f"Vetor Original (v): {v}")
print(f"Matriz da Base C (colunas):\n{P_C}")
print(f"\nSolução encontrada (Coordenadas na Base C):")
print(f"d1 (peso de c1): {d1:.4f} (Fração exata: 7/3)")
print(f"d2 (peso de c2): {d2:.4f} (Fração exata: 2/3)")
# Verificação: Reconstruindo o vetor
v_reconstruido = d1 * c1 + d2 * c2
print(f"\nVerificação (d1*c1 + d2*c2): {v_reconstruido}")
# 4. Visualização Gráfica
plt.figure(figsize=(5, 5))
origin = [0], [0]
# a) Plotar os vetores da Nova Base (para referência)
plt.quiver(*origin, c1[0], c1[1], color='blue', scale=1, scale_units='xy',
angles='xy', label=f'Base c1 {c1}')
```

```python
import numpy as np
import matplotlib.pyplot as plt
# 1. Definição do Vetor e da Nova Base (conforme o texto)
v = np.array([4, 3])
# Vetor no sistema padrão
c1 = np.array([2, 1])   # Primeiro vetor da nova base
c2 = np.array([-1, 1])  # Segundo vetor da nova base
# 2. Montagem da Matriz de Mudança de Base
# A matriz P_C deve ter os vetores da base C como COLUNAS
#P_C = [[2, -1], [1, 1]]
P_C = np.column_stack((c1, c2))
# 3. Resolução do Sistema Linear
# Queremos encontrar d (vetor de coordenadas novas) tal que: P_C * d = v
d = np.linalg.solve(P_C, v)
d1, d2 = d
# Exibição dos resultados numéricos
print("=== Resultados da Mudança de Base ===")
print(f"Vetor Original (v): {v}")
print(f"Matriz da Base C (colunas):\n{P_C}")
print(f"\nSolução encontrada (Coordenadas na Base C):")
print(f"d1 (peso de c1): {d1:.4f} (Fração exata: 7/3)")
print(f"d2 (peso de c2): {d2:.4f} (Fração exata: 2/3)")
# Verificação: Reconstruindo o vetor
v_reconstruido = d1 * c1 + d2 * c2
print(f"\nVerificação (d1*c1 + d2*c2): {v_reconstruido}")
# 4. Visualização Gráfica
plt.figure(figsize=(5, 5))
origin = [0], [0]
# a) Plotar os vetores da Nova Base (para referência)
plt.quiver(*origin, c1[0], c1[1], color='blue', scale=1, scale_units='xy',
angles='xy', label=f'Base c1 {c1}')
plt.quiver(*origin, c2[0], c2[1], color='green', scale=1, scale_units='xy',
angles='xy', label=f'Base c2 {c2}')
# b) Plotar o "Caminho" usando as novas coordenadas
# Passo 1: Andar d1 vezes na direção de c1
passo1 = d1 * c1
plt.quiver(*origin, passo1[0], passo1[1], color='blue', scale=1,
scale_units='xy', angles='xy', alpha=0.3, width=0.015)
# Passo 2: A partir de onde paramos, andar d2 vezes na direção de c2
passo2 = d2 * c2
plt.quiver(passo1[0], passo1[1], passo2[0], passo2[1], color='green', scale=1,
scale_units='xy', angles='xy', alpha=0.3, width=0.015)
# c) Plotar o Vetor Original (Destino)
plt.quiver(*origin, v[0], v[1], color='red', scale=1, scale_units='xy',
angles='xy', label=f'Vetor v {v}', width=0.010)
# Linhas pontilhadas para fechar o paralelogramo visualmente
plt.plot([0, passo1[0]], [0, passo1[1]], 'b--', alpha=0.4, label=f'd1 ({d1:.2f}) * c1')
plt.plot([passo1[0], v[0]], [passo1[1], v[1]], 'g--', alpha=0.4, label=f'd2 ({d2:.2f}) * c2')
# Ajustes do Gráfico
plt.xlim(-2, 5)
plt.ylim(-1, 4)
plt.grid(True, linestyle='--', alpha=0.5)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.legend(loc='upper left')
plt.title(f'Mudança de Base: v = {d1:.2f}*c1 + {d2:.2f}*c2')
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```
