```python
import numpy as np

u = np.array([4, 3, -1])
v = np.array([1, 0, 2])

soma = u + v
print(f"A soma dos vetores é: {soma}")

norma_v = np.linalg.norm(v)
print(f"Norma de v: {norma_v: .4f}")

distancia = np.linalg.norm(u - v)
print(f"Distância {distancia}")

produto_interno = np.dot(u, v)
print(f"Produto interno (dot): {produto_interno}")

produto_interno2 = u @ v
print(f"Produto interno (@)? {produto_interno2}")

produto_manual = np.sum(u * v)
print(f"Produto interno (manual): {produto_manual}")
```
